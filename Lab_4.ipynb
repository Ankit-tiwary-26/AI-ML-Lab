```python
import heapq

heuristic = {
    "A": 2,
    "B": 2,
    "C": 1,
    "D": 3,
    "E": 2,
    "F": 1,
    "H": 3,
    "G": 0
}

level = {
    "A": 0,
    "B": 1,
    "C": 1,
    "D": 2,
    "E": 2,
    "F": 3,
    "H": 3,
    "G": 2
}

graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "E"],
    "D": ["B", "E", "F"],
    "E": ["C", "D", "H"],
    "F": ["D", "G"],
    "H": ["E"],
    "G": []
}

def move_cost(current, neighbor):
    if level[neighbor] > level[current]:
        return 1
    elif level[neighbor] == level[current]:
        return 2
    else:
        return 3

def a_star(start, goal):
    pq = []
    heapq.heappush(pq, (heuristic[start], start))

    came_from = {}
    g_cost = {start: 0}

    explored_order = []

    while pq:
        f, current = heapq.heappop(pq)

        explored_order.append(current)

        if current == goal:
            break

        for neighbor in graph[current]:
            new_g = g_cost[current] + move_cost(current, neighbor)

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                came_from[neighbor] = current
                new_f = new_g + heuristic[neighbor]

                heapq.heappush(pq, (new_f, neighbor))
    path = []
    node = goal
    while node in came_from:
        path.append(node)
        node = came_from[node]
    path.append(start)
    path.reverse()

    return explored_order, path, g_cost.get(goal, None)

start_node = "A"
goal_node = "G"

explored, path, cost = a_star(start_node, goal_node)

print("A* Search - Nodes Explored Order:")
print(" -> ".join(explored))

print("\nFinal Path Found:")
print(" -> ".join(path))

print("\nTotal Cost of Path:", cost)
```

```python
import heapq

heuristic = {
    "A": 2,
    "B": 2,
    "C": 1,
    "D": 3,
    "E": 2,
    "F": 1,
    "H": 3,
    "G": 0
}

level = {
    "A": 0,
    "B": 1,
    "C": 1,
    "D": 2,
    "E": 2,
    "F": 3,
    "H": 3,
    "G": 2
}

graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "E"],
    "D": ["B", "E", "F"],
    "E": ["C", "D", "H"],
    "F": ["D", "G"],
    "H": ["E"],
    "G": []
}

def move_cost(current, neighbor):
    if level[neighbor] > level[current]:
        return 1
    elif level[neighbor] == level[current]:
        return 2
    else:
        return 3

def greedy_best_first(start, goal):
    pq = []
    heapq.heappush(pq, (heuristic[start], start))

    came_from = {}
    visited = set()

    selected_order = []

    while pq:
        h, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)
        selected_order.append(current)

        if current == goal:
            break

        for neighbor in graph[current]:
            if neighbor not in visited:
                came_from[neighbor] = current
                heapq.heappush(pq, (heuristic[neighbor], neighbor))

    path = []
    node = goal
    while node in came_from:
        path.append(node)
        node = came_from[node]
    path.append(start)
    path.reverse()

    total_cost = 0
    for i in range(len(path) - 1):
        total_cost += move_cost(path[i], path[i + 1])

    return selected_order, path, total_cost
start_node = "A"
goal_node = "G"

selected, path, cost = greedy_best_first(start_node, goal_node)

print("Greedy Best First Search - Nodes Selected Order:")
print(" -> ".join(selected))

print("\nFinal Path Found:")
print(" -> ".join(path))

print("\nTotal Cost of Path:", cost)
```
